```python
# /Volumes/workspace/default/my_volume/flight_data.csv

df= spark.read\
    .option('header',True)\
        .option('inferschema',True)\
            .csv('/Volumes/workspace/default/my_volume/flight_data.csv')

df.show(5)
```

```python
#select Transformation----

df.select("DEST_COUNTRY_NAME").distinct().show()
print(df.select("DEST_COUNTRY_NAME").distinct().count())
df.select("ORIGIN_COUNTRY_NAME").distinct().show()
df.select("DEST_COUNTRY_NAME","ORIGIN_COUNTRY_NAME","count").show()
flight_df= df.select("DEST_COUNTRY_NAME","count")
flight_df.show()
```

```python
#Filter Transformation-----

flight_df=df.filter(df["count"]>100)
flight_df.show()
flight_df=df.filter(df["count"]==1)
flight_df.show()
flight_df=df.filter(df["DEST_COUNTRY_NAME"]=="United States")
flight_df.show()
flight_df=df.filter(df["count"]>100).select("DEST_COUNTRY_NAME","count")
flight_df.show(5)
```

```python
#where transformation----
df.where(df["count"]>100).show()
df.where(df["ORIGIN_COUNTRY_NAME"]=="India").show()
```

```python
#Withcolumn transformation----
from pyspark.sql.functions import col,lit,when,upper,length,current_timestamp,current_date
df.withColumn("New_count",col("count")).show()
df.withColumn("count",col("count")*10).show()
df.withColumn("Double_Count",col("count")*2).show(10)
df.withColumn("Source_system",lit("Flight_csv")).show(10)
df.withColumn("flight_status",when(col("count")>=100,"Busy-Route").
              otherwise("Normal-Route")).show(10)
df.withColumn("Route-size",when(col("count")>=200,"High-Traffic").
              otherwise("Low-Traffic")).show(10)
df.withColumn("Traffic-Level",when(col("count")>=200,"High-Traffic").
              when(col("count")>=100,"Medium-Traffic").
              otherwise("Low-Traffic")).show(20)
df.withColumn("DEST_COUNTRY_NAME",upper(col("DEST_COUNTRY_NAME"))).show(10)

df.withColumn("Dest_length",length(col("ORIGIN_COUNTRY_NAME"))).show(10)

df.withColumn("updated_date",current_timestamp()).show(10)

df.withColumn("flight_status",when(col("count")>=100,"Busy-Route").
              otherwise("Normal-Route")).withColumn("Load_timestamp",current_timestamp()).withColumn("Load_date",current_date()).show(10)


```
