```python
# /Volumes/workspace/default/my_volume/flight_data.csv

df= spark.read\
    .option('header',True)\
        .option('inferschema',True)\
            .csv('/Volumes/workspace/default/my_volume/flight_data.csv')

df.show(5)
```

```python
#select Transformation----

df.select("DEST_COUNTRY_NAME").distinct().show()
print(df.select("DEST_COUNTRY_NAME").distinct().count())
df.select("ORIGIN_COUNTRY_NAME").distinct().show()
df.select("DEST_COUNTRY_NAME","ORIGIN_COUNTRY_NAME","count").show()
flight_df= df.select("DEST_COUNTRY_NAME","count")
flight_df.show()
```

```python
#Filter Transformation-----

flight_df=df.filter(df["count"]>100)
flight_df.show()
flight_df=df.filter(df["count"]==1)
flight_df.show()
flight_df=df.filter(df["DEST_COUNTRY_NAME"]=="United States")
flight_df.show()
flight_df=df.filter(df["count"]>100).select("DEST_COUNTRY_NAME","count")
flight_df.show(5)
```

```python
#where transformation----
df.where(df["count"]>100).show()
df.where(df["ORIGIN_COUNTRY_NAME"]=="India").show()
```

```python
#Withcolumn transformation----
from pyspark.sql.functions import col,lit,when,upper,length,current_timestamp,current_date
df.withColumn("New_count",col("count")).show()
df.withColumn("count",col("count")*10).show()
df.withColumn("Double_Count",col("count")*2).show(10)
df.withColumn("Source_system",lit("Flight_csv")).show(10)
df.withColumn("flight_status",when(col("count")>=100,"Busy-Route").
              otherwise("Normal-Route")).show(10)
df.withColumn("Route-size",when(col("count")>=200,"High-Traffic").
              otherwise("Low-Traffic")).show(10)
df.withColumn("Traffic-Level",when(col("count")>=200,"High-Traffic").
              when(col("count")>=100,"Medium-Traffic").
              otherwise("Low-Traffic")).show(20)
df.withColumn("DEST_COUNTRY_NAME",upper(col("DEST_COUNTRY_NAME"))).show(10)

df.withColumn("Dest_length",length(col("ORIGIN_COUNTRY_NAME"))).show(10)

df.withColumn("updated_date",current_timestamp()).show(10)

df.withColumn("flight_status",when(col("count")>=100,"Busy-Route").
              otherwise("Normal-Route")).withColumn("Load_timestamp",current_timestamp()).withColumn("Load_date",current_date()).show(10)


```

```python
#drop transformation---
df.drop("count").show(10)
df.drop("DEST_COUNTRY_NAME").show(10)
df.drop("DEST_COUNTRY_NAME","ORIGIN_COUNTRY_NAME").show(10)
```

```python
#withColumnRenamed transformation---
df.withColumnRenamed("DEST_COUNTRY_NAME","Destination").show(10)
df.withColumnRenamed("count","Flight_Count").show(10)
df.withColumnRenamed("DEST_COUNTRY_NAME","DESTINATION").withColumnRenamed("ORIGIN_COUNTRY_NAME","ORIGIN").show(10)
```

```python
#distinct transformation
print('The overall count is ',df.count())
print('the distinct count is',df.distinct().count())
df.select("DEST_COUNTRY_NAME").distinct().show()
print('The unique destination country are ',df.select("DEST_COUNTRY_NAME").distinct().count())
```

```python
#dropDuplicate transformation---
df.select("DEST_COUNTRY_NAME").dropDuplicates().show()
df.dropDuplicates(["DEST_COUNTRY_NAME","ORIGIN_COUNTRY_NAME"]).show()
```

```python
#groupby transfromation---
from pyspark.sql.functions import avg,sum,count,max,min

df.groupBy("DEST_COUNTRY_NAME").count().show(10)
df.groupby("DEST_COUNTRY_NAME").avg("count").show()
df.groupBy("DEST_COUNTRY_NAME").max("count").show()
df.groupBy("DEST_COUNTRY_NAME").agg(
    sum("count").alias("Total flight count"),
    avg("count").alias("Average flight count"),
    max("count").alias("Maximum flight count")
).show()
df.groupby("ORIGIN_COUNTRY_NAME").count().show()
df.groupBy("ORIGIN_COUNTRY_NAME").agg(
    sum("count").alias("Total flight count"),
    avg("count").alias("Average flight count"),
    max("count").alias("Maximum flight count")
).show()
df.groupby("DEST_COUNTRY_NAME","ORIGIN_COUNTRY_NAME").count().show()
```

```python
#orderby transformation---
from pyspark.sql.functions import col

df.orderBy("count").show(5)
df.orderBy(col("count").desc()).show(5)
df.orderBy("DEST_COUNTRY_NAME",col("count").desc()).show(10)
df.orderBy(col("count").desc()).limit(10).show()
```

```python
#limit tranfromations---
from pyspark.sql.functions import col

df.limit(10).show()
df.filter(df["count"]>100).limit(50).show()
df.orderBy(df["count"].desc()).limit(10).show()
df.orderBy("count").limit(10).show()
df.filter(df["ORIGIN_COUNTRY_NAME"]=="India").orderBy(col("count").desc()).limit(5).show()
```
