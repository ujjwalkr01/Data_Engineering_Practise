```python
data = [
    (101, "Rahul", "Delhi", 50000, "rahul@gmail.com", None),
    (102, "Amit", None, 60000, None, "9876543210"),
    (103, None, "Mumbai", None, "ravi@gmail.com", None),
    (104, "Priya", None, 70000, None, "9123456789"),
    (105, "Neha", "Pune", None, None, None)
]

columns = [
    "customer_id",
    "name",
    "city",
    "salary",
    "primary_email",
    "secondary_phone"
]

customer_df = spark.createDataFrame(data, columns)
customer_df.show()
```

```python
res1_df=customer_df.filter(customer_df.city.isNull())
res1_df.show()

res2_df=customer_df.filter(customer_df.city.isNotNull())
res2_df.show()
```

```python
res3_df=customer_df.fillna({"city":"Unknown"})
res3_df.show()

res4_df=customer_df.fillna({"salary":0})
res4_df.show()
```

```python
res5_df=customer_df.dropna(subset=["customer_id"])
res5_df.show()
```

```python
from pyspark.sql.functions import col,coalesce

res6_df=customer_df.withColumn(
    "email",
    coalesce(col("primary_email"), col("secondary_phone"))
)
res6_df.show()
```
