```python
flight_data = [
    ("USA", "India", 100),
    ("USA", "Canada", 80),
    ("USA", "Mexico", 80),
    ("USA", "UK", 60),

    ("India", "USA", 120),
    ("India", "Canada", 90),
    ("India", "UK", 90),
    ("India", "Australia", 50),

    ("Canada", "USA", 110),
    ("Canada", "India", 70),
    ("Canada", "UK", 70),
    ("Canada", "France", 40)
]

flight_columns = [
    "ORIGIN_COUNTRY_NAME",
    "DEST_COUNTRY_NAME",
    "count"
]

flight_df = spark.createDataFrame(
    flight_data,
    flight_columns
)

flight_df.show()
```

```python
from pyspark.sql.functions import col

res1_df=flight_df.filter(col("count")>100).select("DEST_COUNTRY_NAME","count")
res1_df.explain()
```

```python
res2_df=flight_df.filter(col("count")>100).groupBy("ORIGIN_COUNTRY_NAME").count()
res2_df.explain()
```

```python
res3_df = flight_df.orderBy(col("count").desc())

res3_df.explain()
```
