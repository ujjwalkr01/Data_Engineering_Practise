```python
flight_data = [
    ("USA", "India", 100),
    ("USA", "Canada", 80),
    ("USA", "Mexico", 80),
    ("USA", "UK", 60),

    ("India", "USA", 120),
    ("India", "Canada", 90),
    ("India", "UK", 90),
    ("India", "Australia", 50),

    ("Canada", "USA", 110),
    ("Canada", "India", 70),
    ("Canada", "UK", 70),
    ("Canada", "France", 40)
]

flight_columns = [
    "ORIGIN_COUNTRY_NAME",
    "DEST_COUNTRY_NAME",
    "count"
]

flight_df = spark.createDataFrame(
    flight_data,
    flight_columns
)

flight_df.show()
```

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number,rank,desc,col,dense_rank

windowSpec=Window.partitionBy("ORIGIN_COUNTRY_NAME").orderBy(col("count").desc())

res1_df=flight_df.withColumn("row_num", 
                             row_number().over(windowSpec))
res1_df.show()

res2_df=flight_df.withColumn("rank",rank().over(windowSpec))
res2_df.show()

res3_df=flight_df.withColumn("dense_rank",dense_rank().over(windowSpec))
res3_df.show()
```

```python
res5_df=flight_df.withColumn("row_num",
                             row_number().over(Window.partitionBy("ORIGIN_COUNTRY_NAME").orderBy(col("count").desc()))).filter(col("row_num")<=2)
res5_df.show()

res6_df=flight_df.withColumn("rank",dense_rank().over(windowSpec)).filter(col("rank")<=2)
res6_df.show()
```

```python
res7_df=flight_df.withColumn("row__number",row_number().over(Window.orderBy(col("count").desc())))
res7_df.show()

res8_df=flight_df.withColumn("rnk",rank().over(windowSpec)).filter(col("rnk")==1)
res8_df.show()
```
