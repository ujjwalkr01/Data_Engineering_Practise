```python
data1 = [
    (1, "Rahul", 50000),
    (2, "Amit", 60000)
]

columns1 = ["id", "name", "salary"]

df1 = spark.createDataFrame(data1, columns1)
df1.show()

data2=[
    (3, "Ravi", 55000),
    (4, "Priya", 70000)
]

columns2=["id","name","salary"]

df2=spark.createDataFrame(data2, columns2)
df2.show()

data3 = [
    (80000, 5, "Neha"),
    (90000, 6, "Karan")
]

columns3 = ["salary", "id", "name"]

df3 = spark.createDataFrame(data3, columns3)
df3.show()

data4 = [
    (7, "Suresh"),
    (8, "Anita")
]

columns4 = ["id", "name"]

df4 = spark.createDataFrame(data4, columns4)
df4.show()
```

```python
res1_df=df1.union(df2)
res1_df.show()

#removes duplicates
res2_df=df1.union(df2).distinct()
res2_df.show()

#positioning the correct columns according to the columnName
res3_df=df1.unionByName(df3)
res3_df.show()
```

```python
#provide values to missing columns:
res5_df=df1.unionByName(df4, allowMissingColumns=True)
res5_df.show()
```
