```python
emp_details=spark.read.format("csv")\
    .option("header",True)\
        .option("inferschema",True)\
            .load("/Volumes/workspace/default/my_volume/emp_details.csv")

emp_details.show()
```

```python
emp_details.write.format("csv").option("header", True).option(
    "mode", "overwrite"
).option("path", "/Volumes/workspace/default/my_volume/emp_details_write").save()
```

```python
dbutils.fs.ls("/Volumes/workspace/default/my_volume/emp_details_write")
```

```python
emp_details.repartition(3).write.format("csv").option("header", True).option(
    "mode", "overwrite"
).option("path", "/Volumes/workspace/default/my_volume/emp_details_write_repartition").save()
```

```python
dbutils.fs.ls("/Volumes/workspace/default/my_volume/emp_details_write_repartition")
```
