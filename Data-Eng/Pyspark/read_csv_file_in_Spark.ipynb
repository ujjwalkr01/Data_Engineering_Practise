```python
from pyspark.sql.types import StructType,StructField,StringType,IntegerType

my_schema=StructType( [
                        StructField('DEST_COUNTRY_NAME',StringType(),True),
                        StructField('ORIGIN_COUNTRY_NAME',StringType(),True),
                        StructField('Count',IntegerType(),True)
                       ]
                      )
```

```python
flight_df=spark.read.format('csv')\
              .option('header','false')\
                .option('inferschema','false')\
                    .schema(my_schema)\
                        .option('mode','FAILFAST')\
                             .load('/Volumes/workspace/default/my_volume/flight_data.csv')

flight_df.show(5)
```

```python
flight_df = (
    spark.read.format("csv")
    .option("header", "false")
    .option("skipRows", 1)
    .option("inferschema", "false")
    .schema(my_schema)
    .option("mode", "PERMISSIVE")
    .load("/Volumes/workspace/default/my_volume/flight_data.csv")
)

flight_df.show(5) # here header will not print as we have use SkipRows snd skip the first row
```

```python
from pyspark.sql.types import StructType,StructField,StringType,IntegerType

emp_schema_df=StructType( [
                         StructField('Id',IntegerType(),True),
                         StructField('Name',StringType(),True),
                         StructField('Age',IntegerType(),True),
                         StructField('Salary',IntegerType(),True),
                         StructField('Address',StringType(),True),
                         StructField("Nominee",StringType(),True)
                       ])

print(emp_schema_df.json())
#/Volumes/workspace/default/my_volume/emp_Corrupted_data.csv

```

```python
emp_corruptedRec_df=spark.read.format('csv')\
    .option('header',True)\
        .option('inferschema',True)\
            .schema(emp_schema_df)\
                .option('mode','PERMISSIVE')\
                    .load('/Volumes/workspace/default/my_volume/emp_Corrupted_data.csv')

emp_corruptedRec_df.show()
```

```python
emp_corruptedRec_df=spark.read.format('csv')\
    .option('header',True)\
        .option('inferschema',True)\
            .schema(emp_schema_df)\
                .option('mode','DROPMALFORMED')\
                    .load('/Volumes/workspace/default/my_volume/emp_Corrupted_data.csv')

emp_corruptedRec_df.show()
```

```python
from pyspark.sql.types import *

emp_schemaCorupt_df=StructType( [
                         StructField('Id',IntegerType(),True),
                         StructField('Name',StringType(),True),
                         StructField('Age',IntegerType(),True),
                         StructField('Salary',IntegerType(),True),
                         StructField('Address',StringType(),True),
                         StructField("Nominee",StringType(),True),
                         StructField('corrupt_rec',StringType(),True)
                       ])

print(emp_schemaCorupt_df.json())
#/Volumes/workspace/default/my_volume/emp_Corrupted_data.csv

```

```python
emp_corruptedRecCaptured_df = (
    spark.read.format("csv")
    .option("header", True)
    .option("inferschema", True)
    .option("mode", "PERMISSIVE")
    .option("columnNameOfCorruptRecord", "corrupt_rec")
    .schema(emp_schemaCorupt_df)
    .load("/Volumes/workspace/default/my_volume/emp_Corrupted_data.csv")
)

emp_corruptedRecCaptured_df.show(truncate=False)#to print whole bad records
```
