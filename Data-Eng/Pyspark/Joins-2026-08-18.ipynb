```python
df= spark.read\
    .option('header',True)\
        .option('inferschema',True)\
            .csv('/Volumes/workspace/default/my_volume/flight_data.csv')

df.show(5)
```

```python
employee_data = [
    (1, "Rahul", 10, 50000),
    (2, "Amit", 20, 60000),
    (3, "Ravi", 30, 55000),
    (4, "Priya", 40, 70000)
]

employee_columns = ["employee_id", "name", "dept_id", "salary"]

employee_df = spark.createDataFrame(
    employee_data,
    employee_columns
)

employee_df.show()
```

```python
department_data = [
    (10, "IT"),
    (20, "HR"),
    (30, "Finance"),
    (50, "Marketing")
]

department_columns = ["dept_id", "department"]

department_df = spark.createDataFrame(
    department_data,
    department_columns
)

department_df.show()
```

```python
InnerJoin_df= employee_df.join(department_df,"dept_id","inner").show()

LeftJoin_df=employee_df.join(department_df,"dept_id","left").show()

RightJoin_df=employee_df.join(department_df,"dept_id","right").show()

FullOuter_df=employee_df.join(department_df,"dept_id","full_outer").show()
```

```python
from pyspark.sql.functions import col

e=employee_df.alias("e")
d=department_df.alias("d")

result_df=e.join(d,
                 col("e.dept_id")==col("d.dept_id"),
                 "inner").select(col("e.employee_id"),col("e.name"),col("d.department"),col("e.salary")).show()
```
